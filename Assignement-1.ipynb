```python
BMI = float(input("Enter the BMI(Body Mass Index): "))
if(BMI <= 18.5):
    print("Underweight")

elif(BMI>18.5 and BMI <=24.9):
    print("Normal")

elif(BMI>=25 and BMI <= 29.9):
    print("Overweight")

elif(BMI>=30 and BMI <=34.9):
    print("Obese")

elif(BMI>=35 and BMI <= 39.9):
    print("Severely Obese")

else:
    print("Mobidly Obese")
```
