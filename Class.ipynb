```python
class OddEven:
    def OddEven():
        Num = int(input("Enter a number:"))
        if(Num%2==0):
       # print("The Number",Num,"is Even")
            message = "The Number " + str(Num) + " is Even"
        else:
        #print("The Number",Num,"is Odd")
            message = "The Number " + str(Num) + " is Odd"
        return message

print(OddEven.OddEven())
```

```python
class FindPercentage:
    def percentage(marks):
        total=0
        for i in marks:
            total+=i
        percentage = total/5
        return percentage
marks = []
for i in range(5):
    mark = int(input("Enter mark for Subject "+str(i+1)+": "))
    marks.append(mark)

result = FindPercentage.percentage(marks)
print (result)   
```
