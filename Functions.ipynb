```python
def OddEven():
    Num = int(input("Enter a number:"))
    if(Num%2==0):
       # print("The Number",Num,"is Even")
        message = "The Number " + str(Num) + " is Even"
    else:
        #print("The Number",Num,"is Odd")
        message = "The Number " + str(Num) + " is Odd"
    return message

oddEven = OddEven()
print(oddEven)
```

```python
def MarriageEligibility(age, gender):
    if gender.lower() == "male" and age >=21:
        return "Eligible for marriage"
    else:
        return "Not Eligible for marriage"
    if gender.lower() == "female" and age>=18:
        return "Eligible for marriage"
    else:
        return "Not Eligible for marriage"

Age = int(input("Enter your Age:"))
Gender = input("Enter your Gender")
print(MarriageEligibility(Age, Gender))
```
